```python
import pandas as pd

data = {
    "existing_column": [60, 45, 70, 85, 55, 40],
    "another_column": [80, 90, 50, 70, 60, 95],
    "numeric_column": [100, 80, 90, 120, 110, 75],
    "category_column": ["A", "B", "A", "C", "B", "C"],
    "column1": ["Apple", "Banana", "Mango", "Orange", "Grapes", "Pineapple"],
    "column2": [10, 20, 30, 40, 50, 60]
}

df = pd.DataFrame(data)

print("First 5 Rows")
print(df.head())

print("\nLast 5 Rows")
print(df.tail())

print("\nInformation")
df.info()

print("\nSummary Statistics")
print(df.describe())

df["new_column"] = df["existing_column"] * 2

series = df["existing_column"]
print("\nSeries + 10")
print(series + 10)

filtered_df = df[(df["existing_column"] > 50) & (df["another_column"] < 100)]
print("\nFiltered Data")
print(filtered_df)

grouped = df.groupby("category_column")["numeric_column"].mean()
print("\nGrouped Mean")
print(grouped)

df_sorted = df.sort_values(by="numeric_column", ascending=False)
print("\nSorted Data")
print(df_sorted)

masked_df = df[df["numeric_column"] > df["numeric_column"].median()]
print("\nMasked Data")
print(masked_df)

df.drop_duplicates(inplace=True)

subset_df = df[["column1", "column2"]]

subset_df.to_csv("filtered_data.csv", index=False)

print("\nTotal Sum")
print(df["numeric_column"].sum())

print("\nMean")
print(df["numeric_column"].mean())

print("\nStandard Deviation")
print(df["numeric_column"].std())
```
